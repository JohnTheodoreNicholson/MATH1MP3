```python
import math
def Bissection(f, a, b):
    '''
    Solves the equation f(x) = 0 using the bissection 
    algorithm given a bracketting interval (a,b)
    '''
    counter = 0
    while counter < 50:
        m = (a + b) / 2
        if f(m) == 0:
            return m
        elif f(b) * f(m) < 0:
            a = m
        else:
            b = m
        print(f"Iteration {counter} m = {m}, (a,b) = ({a}, {b}), f(a) = {f(a)}, f(b) = {f(b)}") 
        counter += 1
    return m

def f(x):
    return math.exp(x)-1

m = Bissection(f, -10, 21)
print(m)
    
```

```python
def is_palindrome(n):
    is_palindromic = False
    string_n = str(n)
    string_reversed = string_n[::-1]
    if string_n == string_reversed:
        is_palindromic = True
    return is_palindromic
```

```python
def sum(A):
    n = len(A)
    grandsum = 0
    for row_idx in range(n):
        for col_idx in range(n):
            grandsum += A[row_idx][col_idx]
    return grandsum
```
