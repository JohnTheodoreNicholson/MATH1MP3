```python
import numpy as np

def local_max(A):
    indices = []
    for idx, el in np.ndenumerate(A):
        if idx[0] == 0 and el > A[idx[0]+1]:
            indices += [idx]
        elif idx[0] == len(A) -1 and el > A[idx[0]-1]:
            indices += [idx]
        elif el > A[idx[0]-1] and el > A[idx[0]+1]:
            indices += [idx]
    return indices

A = np.array([7,2,3,1,5,7])
print(local_max(A))
```

```python
import math
def Bissection(f, a, b):
    '''
    Solves the equation f(x) = 0 using the bissection 
    algorithm given a bracketting interval (a,b)
    '''
    counter = 0
    while counter < 50:
        m = (a + b) / 2
        if f(m) == 0:
            return m
        elif f(b) * f(m) < 0:
            a = m
        else:
            b = m
        print(f"Iteration {counter} m = {m}, (a,b) = ({a}, {b}), f(a) = {f(a)}, f(b) = {f(b)}") 
        counter += 1
    return m

def f(x):
    return math.exp(x)-1

m = Bissection(f, -10, 21)
print(m)
    
```

```python
def is_palindrome(n):
    is_palindromic = False
    string_n = str(n)
    string_reversed = string_n[::-1]
    if string_n == string_reversed:
        is_palindromic = True
    return is_palindromic
```
