Another classical calculus concept is Taylor polynomials: Let $f$ be infinitely differentiable at a point $x_0$ and $n$ be a non-negative integer. The $n^\mathrm{th}$ order Taylor polynomial of $f$ at $x_0$ is:
$$
T_{n, x_0}(x) = \sum_{i=0}^n f^{(i)}(x_0)\frac{(x-x_0)^i}{i!},
$$
where $f^{(i)}$ is the $i^\mathrm{th}$ derivative of $f$.

Using the `numpy.polynomial` class, write a function `Taylor(x,n)` returning the Taylor polynomial or order $n$ of $f(x) = \sin x$ at some $x_0$.

### Question 5: Euclid's algorithm (4 marks)

Euclid's algorithm is a classical algorithm to compute the greatest common divisor (GCD) of two integers:

* let $a$ and $b$ be given
* set $a_0 = a$ and $b_0 = b$
* repeat:
    - $a_i = b_i  q_i + r_i$ with $r_i < b_i$.
    - if $r_0 = 0$ then return $GCD(a, b) = b_i$.
    - $a_{i+1} \leftarrow b_i$
    - $b_{i+1} \leftarrow r_i$.
* until $r_i = 0$

Note: since $b_{i+1} = r_i < b_i$, this algorithm will always converge in at most $b$ iterations (much less in practice)!
