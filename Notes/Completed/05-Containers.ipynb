## Containers
The two largest collections of types in Python are numerical types and **containers**. Containers are types that may contain other objects. They share methods to refer to some of their content, or perform operations on them.
Typical operations are 
* indexing `x[i]`
* query `'M' in '1MP3'` 
* extension, concatenation, restriction, ...

Indexing can be used to extract part of a string. Given a string `s`, `s[a:b]` denotes the string consistings of characters at position `a` to `b` (excluded).

`len(s)` return the length of `s`, i.e. number of characters in `s`

We have already seen one container type so far, **strings**!

```python
s = 'test string'
print('s:    ', s)
print('s[-1]: ', s[-1])
print('s[-1]: ', s[-2])
print('s[0]: ',s[0])
print('s[0:5:2]: ',s[0:5:2])
```

Recall, we have previously discussed how strings are a **homogeneous** container. That is, any object within the string will also have type string.

```python
# an example
s = '12abc'
print(type(s[0]))
print(type(s[3]))
```

However, Python does still "know" that these objects have different type.

```python
# this works
print(int(s[0]))
```

```python
# but this doesn't
print(int(s[3]))
```

### A common error

The statement `s[i]` asks Python to fetch the value that lives at index `i` in the string container. It is assumed that this index is a valid one, that is, one that lives within the container. If it isn't, we will get an error.

```python
s = 'a longer string'
print(len(s))
```

```python
# this is fine
print(s[14])
```

```python
# but this is not
print(s[15])
```

#### String slicing

```python
print(s)
print('s[0:3]:  ', s[0:3])
print('s[3:-1]: ', s[3:]-1)
print(len(s))
```

If `a` is omitted, `s[:b]` is to be understood as all positions up to (excluding) `b`.
If `b` is omitted, `s[a:]` is to be understood as all starting with `a`.


```python
print('s[:3]: ',s[:3])
print('s[:3]: ',s[0:3])
print('s[3:]: ',s[3:])
print(s[len(s)-1])
```

Optionally, one can also specify a stride: `s[a:b:c]` is a string containing characters at indices `a`, `a+c`, `a+2*c`, ... `b` in `s`.

```python
numbers = '0123456789'
print(numbers)
print('numbers[0:10:2]: ', numbers[0:10:2])
print('numbers[1:10:2]: ', numbers[1:10:2])
print('numbers[1:10:3]: ', numbers[1:10:3])
```

Note that the stride can be negative, in which case one needs to get `a>b`

```python
print('numbers[10:0:-1]: ', numbers[10:0:-1])
```

Note that the rules for omitted bounds still apply when a stride is specified:

```python
print('numbers[0:10:2]: ', numbers[0:10:2])
print('numbers[:10:2]:  ', numbers[:10:2])
print('numbers[::2]:    ', numbers[0::2])
```

```python
msg = "this is a test"
# You cannot capitalize the first character this way;
msg[0] = "T"
```

### Lists

A *list* is an *ordered* collection of objects (i.e. `[1,2]` and `[2,1]` are not equal).

* Lists are defined using square brackets ('[]') or the `list()` operator. 
* Lists can be indexed.
* List are mutable


```python
l = [0,1,2,3,4,5,6]
print(l)
print(len(l))
print(l[0])
print(l[-1])
print(l[0:4])
```

```python
l = [0,1,2,3,4,5,6]
print('l:      ', l)
print('l[0]:   ', l[0])
print('l[1:3]: ', l[1:3])
print('l[::-2]: ', l[::-2])
```

```python
# strings are immutable
s = "1,2,3,4"
s[-1] = "5"
```

```python
# lists are mutable!
s = [1,2,3,4]
print(s)
s[-1] = 5
print(s)
```

Another unusual feature of python is that not all elements in a list must be the same type of objects:

```python
print(type(l[0]))
print(type(l[1]))
print(type(l[-1]))
```

Unlike with strings, in the statement above `l[0]` will have type `int` and `l[1]` will have type `str`.

We can represent matrices in Python with nested lists. That is, the matrix
$$\begin{bmatrix}
1 & 2 & 3\\
4 & 5 & 6\\
7 & 8 & 9
\end{bmatrix}$$
can be represented by the nested list `[[1,2,3],[4,5,6],[7,8,9]]`. 

Using this representation, write code that computes the determinant of a 2 by 2 matrix represented as a nested list. **Recall:** Given a 2 by 2 matrix of the following form,
$$\begin{bmatrix}
a & b\\ c & d
\end{bmatrix}$$
the determinant can be computed as $ad-bc$.

```python
m = [[1,2],[3,4]]
print(m)
print(m[0][0])
a = m[0][0]
b = m[0][1]
c = m[1][0]
d = m[1][1]

det = a*d - b*c
print(det)
```

```python
# here is a 2 by 2 matrix
m = [[1,2],[3,4]]
# write code to compute its determinant!
print(m[0][0])
print(m[1][1])
print(m[0][0]*m[1][1] - m[0][1]*m[1][0]) # formula for 2 by 2 determinant 
```

### Dictionaries

Dictionaries are a convenient way to store *structured* data.
They can be thought of as (unordered) lists whose indexes can be any object (hence do not have a canonical ordering but must be unique), or as collections of pairs `(key:value)`.
Dictionaries use curly brackets `{}`

```python
d = {'a':1, 'b':2, 'c':3}
print('d: ', d)
print("d['a']: ", d['a'])
```

```python
d = {0 : 1, 1:2, 2:3}
L = [1,2,3]
print(d[0])
print(L[0])
```

The values of a dictionary can be any Python object. The keys can be any *immutable* Python object (it wouldn't be a good idea if the key to a certain value could be changed arbitrarily.)

```python
d2 = {1:'abc', (0,1):[1,2,3]}
print(d2[1])
print(d2[(0,1)])
```

### Ranges

Python has a data type along with a clean syntax for creating ordered collections of integers, given a starting point, an end point, and a step size (similar syntactically to list indexing). This type is called a **range**. We provide some examples below.

```python
r = range(1,5,1)
print(f"r = {r}")
print(f"list(r) = {list(r)}")
```

```python
# default start = 0
# default step-size = 1
r = range(6)
print(list(r))
```

```python
# make a range of all of the even integers between 1 and 100
even = range(2,101,2)
print(list(even))
```
