So far, we have seen two root finding algorithms (with some variants), **bisection** and **Newton's method**. Bisection and Newton represent the two main categories of root finding algorithms that exist, namely 
- *bracketing* algorithms, and 
- *interpolation* algorithms. 

Bracketing algorithms, like bisection, take as input an interval upon which $f$ has a root, and aims to iteratively shrink the size of the interval to better approximate the root. Interpolation algorithsm, like Newton's method, take as input an approximation (or a first guess) of the location of a root, and aim to iteratively obtain better approximations.

We will now talk briefly about the possible extensions of each of these algorithms.

```python
def bisection(f ,a : float,b : float,tol = 1e-10, maxIter = 100) -> float | tuple[float,float] :
    if f(a)*f(b)< 0: # if false
        ctr = 0
        # f doesn't change sign on [a,b]
        while abs(a - b) >= tol and ctr <= maxIter:
            c = (a + b)/2 #midpoint of the interval
            if f(a) == 0.0 :
                return f"Found a root at x = {a}."
            elif f(b) == 0.0:
                return f"Found a root a x = {b}."
            elif f(c) == 0.0:
                return f"Found a root a x = {c}."
            elif f(a)*f(c) < 0:
                b = c
            else:
                a = c
            ctr +=1
        return (a,b)

    else:
        return None

# Fixed tolerance for our computation
TOL = 1.0e-12

def Newton(f , fprime, a : float,max_iter = 1000, tol = TOL) -> float:
    err = abs(f(a))
    ctr = 0 # count the number of iterations
    while ctr < max_iter and err > tol:
        #err is still too big or we haven't reached the maximal number of iterations
        a = a - f(a)/fprime(a)
        # new guess for a
        err = abs(f(a))
        ctr += 1
    return a

def secant(f,a,b,eps, nMax = 100):
    # no fprime
    numIter = 0
    err = eps + 1
    err = abs(f(a))
    while numIter < nMax and err > eps:
        numIter += 1
        # approximating fprime(a) by the secant line
        # between a and b
        c = a-f(a) * (b-a) / (f(b) - f(a))
        err = abs(f(c))
        a = b
        b = c
        print(numIter, c, err)
    return c


# Bisection converges linearly (1/2)
# Newton's converges quadratically (2)
# Secant converges "superlinearly" (1.62..)
```

## Fixed point iteration

Consider the following sequence: $x_0 = 0$, $x_{n+1}=cos(x_n)$. What is the behaviour of this sequence?

```python
import numpy as np
x = 0
n = 1000
i = 0
while i<n:
    x = np.cos(x)
    i+=1
    if i%100:
        print(x)
```

We can observe that this sequence of $x_{n+1}= cos(x_n)$ will converge to approximately $0.7390851332151607$. This is telling us that $cos(0.739...)=0.739...$ 

This is called a **fixed point** of cosine.

## Newton is a fixed point algorithm

**Notice:** Let $x_0=a$, $x_{n+1}=x_{n}-f(x_n)/f'(x_n)$ denote the sequence obtained by iterating Newton's method. Notice that Newton's method converges (in the computational sense), when there exists $N\geq 1$, such that $X_{N+1}=X_N$. If we let $g(x)=x-f(x)/f'(x)$, then this is equivalent to saying that Newton's method converges for $f$ when we can solve $g(x)=x$ by fixed point iteration.

## When does fixed point iteration converge?

**Theorem (Fixed Point Convergence Theorem):** Given a function $f(x)$ with continuous first derivative and fixed point $\hat{x}$, if $|f'(\hat{x})|<1$, then there exists a neighborhood of $\hat{x}$ in which fixed point iteration converges for any initial value in the neighborhood.

## Example - Finding roots using fixed point iteration

Given the function 
$$f(x) = x^2+x-1$$

We want to rewrite the equation $f(x)=0$ as an equation of the form $g(x)=x$ for some $g(x)$ closely related to $f(x)$. For the above equation this can be done in the following ways:
$$g_1(x) = (1/x)-1$$
$$g_2(x) = \frac{1}{(x+1)}$$
$$g_3(x) = x^2+2x-1$$
$$g_4(x) = \pm\sqrt{1-x}$$

Then, finding roots of $f(x)$ is equivalent to finding a fixed point of $g_1,\dots,g_4$. 

**Important Note:** Some choices of $g$ are better than others! For example $g_2$ will provide much more robust convergence than $g_3$.

```python
f = lambda x : x**2+x-1

x = np.linspace(-100,100,1000)
y = f(x)

plt.plot(x,y)
plt.show()
```

# Fractals via Iterative Methods

## Example 1 - Newton Fractals

Let $f:\mathbb{C}\rightarrow\mathbb{C}$$ be a function with $n$ roots (for us it is fine to think of $\mathbb{C}=\mathbb{R}^2$). We will draw a plot of the plane, where a point $p\in\mathbb{C}$ gets colour $c_i$ (for $1\leq i\leq n$) if starting Newton's method from $p$ converges to the $i^{th}$ root of $f$. 

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from collections.abc import Callable

# Fixed tolerance for our computation
TOL = 1.0e-12

def Newton(f : Callable[[float],float], fprime, a : float,max_iter = 1000, tol = TOL) -> float:
    err = abs(f(a))
    ctr = 0 # count the number of iterations
    while ctr < max_iter and err > tol:
        #err is still too big or we haven't reached the maximal number of iterations
        a = a - f(a)/fprime(a)
        # new guess for a
        err = abs(f(a))
        ctr += 1
    return a
```

```python
def get_root_index(roots, r):
        try:
            return np.where(np.isclose(roots, r, atol=TOL))[0][0]
        except IndexError:
            roots.append(r)
            return len(roots) - 1
```

```python
def plot_newton_fractal(f, fprime, n=500, domain=(-1, 1, -1, 1)):
    roots = []
    m = np.zeros((n, n))
    xmin, xmax, ymin, ymax = domain
    for ix, x in enumerate(np.linspace(xmin, xmax, n)):
        for iy, y in enumerate(np.linspace(ymin, ymax, n)):
            z0 = x + y * 1j

            r = Newton(f, fprime, z0,)
            ir = get_root_index(roots, r)
            m[iy, ix] = ir
    plt.imshow(m, origin="lower", interpolation = "gaussian")
    plt.axis("off")
    plt.show()
```

```python
plot_newton_fractal(lambda z: z**4 - 1, lambda z: 4 * z**3, n = 500)
```

```python
plot_newton_fractal(lambda z : z**3 - 1, lambda z : 3*z**2, domain = (-5,5,-5,5))
```
