# Type conversion

We have seen thus far that objects in Python have type. Usually, it will only make sense to define operations on objects of uniform type (for example the addition of integers.) However, we have seen that in special cases, Python can allow for operations between objects of different types. 

This is done through *implicit type conversion*.

```python
# example 1
x = 1
print(x, type(x))
x = x/3.
print(x, type(x))
```

```python
# example 2
x = 1
y = 2.6
print(f"x has type {type(x)}")
print(f"y has type {type(y)}")
print(f"x + y has type {type(x+y)}")
```

```python
x = 'an example string'
print(x, type(x))
```

It is possible (but rarely necessary) to force conversion between types using the `int()`, `float()`, or `str()` operations

```python
x = str(42)
print(x, type(x))
x = float(42)
print(x, type(x))
```

```python
x = str(42)
x[0]
```

```python
t = 'an example string'
t[0:2]
```

```python
# this makes no sense
x = int('an example string')
```

```python
# but this does
x = '2.3'
print(x, type(x))
x = float(x)
print(x, type(x))
```

```python
x = 1
x = float(x) + 2.3
print(x, type(x))
```

```python
x = 8.999
#truncation - not rounding
print(int(x))
```

```python
x = 3
x = 'an example string' + str(x) 
print(x, type(x))
```

**Notice:** Even when a certain type conversion is allowed by Python, it may not be advisable. For instance, consider the following example:

```python
x = 2.9999
print(f"x = {x}")
print(f"int(x) = {int(x)}")
```
