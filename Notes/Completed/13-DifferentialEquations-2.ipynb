# Applying Euler's method to disease modelling
## An introduction to the SIR model

One of the most famous models for modelling the dynamics of an epidemic is called the SIR model. This stands for the three groups, this model considers: Susceptible (can still get the disease); Infected (currently has the disease); and Recovered (no longer has the disease, but is assumed to have gained immunity). This can be described by the following system of differential equations:
\begin{align*}
\frac{dS}{dt}&=-\frac{\beta}{N}IS,\\
\frac{dI}{dt}&=\frac{\beta}{N}IS-\gamma I,\\
\frac{dR}{dt}&=\gamma I,
\end{align*}

Here $N$ is the total number of people in the system, $\beta$ is the infection rate constant, and $\gamma$ is the recovery rate constant. Time is usually considered to be in units of days. 

This model was originally introduced by Kermack and McKendrick. 

```python
# Constants for the simulation
days = 365
h = 1e-4 # step-size for Euler's method
total_steps = int(days / h)

# time array
t = np.linspace(0, days, total_steps)

# initializing the arrays we will plot
S = np.zeros_like(t)
I = np.zeros_like(t)
R = np.zeros_like(t)

# Setting initial values
S[0] = 997
I[0] = 3
R[0] = 0
N = S[0]+I[0]+R[0]
gamma = 0.04
beta = 0.0012

for i in range(total_steps-1):
    dS = -(beta/N)*S[i]*I[i]
    dI = (beta/N)*I[i]+S[i] - gamma*I[i]
    dR = gamma*I[i]

    S[i+1] = S[i]+h*dS
    I[i+1] = I[i]+ h*dI
    R[i+1] = R[i]+h*dR

plt.plot(t,S,label="S")
plt.plot(t,I,label="I")
plt.plot(t,R,label="R")
plt.legend()
plt.show()
```

# Discussion

The model we saw above is called a "compartment" model. It gets this name as in the model we cut the population up into mutually exclusive groups (or compartments). The differential equations in our model then describe how the population can flow from one group to another. Compartment models are used in a variety of scientific modeling applications. For instance, most economic models are compartment models.
