## 1. The bissection method

```python
import math
def Bissection(f, a, b):
    '''
    Solves the equation f(x) = 0 using the bissection 
    algorithm given a bracketting interval (a,b)
    '''
    counter = 0
    while counter < 50:
        m = (a + b) / 2
        if f(m) == 0:
            return m
        elif f(a) * f(m) < 0:
            b = m
        else:
            a = m
        print(f"Iteration {counter} m = {m}, (a,b) = ({a}, {b}), f(a) = {f(a)}, f(b) = {f(b)}") 
        counter += 1
    return m

def f(x):
    return math.exp(x)-1

m = Bissection(f, -10, 21)
print(m)
    
```

```python
def Bissection2(f, a, b, epsilon):
    '''
    Solves the equation f(x) = 0 using the bissection 
    algorithm given a bracketting interval (a,b)
    '''
    counter = 0
    while b - a > epsilon:
        m = (a + b) / 2
        if f(m) == 0:
            return m
        elif f(a) * f(m) < 0:
            b = m
        else:
            a = m
        print(f"Iteration {counter} m = {m}, (a,b) = ({a}, {b}), b-a {b-a}") 
        counter += 1
    return m

def f(x):
    return math.exp(x)-1

m = Bissection2(f, -10, 21, 1e-4)
print(m)
```

```python
# Another convergence test:
# iterate until |f(m)| < epsilon 
def Bissection3(f, a, b, epsilon):
    '''
    Solves the equation f(x) = 0 using the bissection 
    algorithm given a bracketting interval (a,b)
    '''
    counter = 0
    m = (a + b) / 2
    while abs( f(m) ) > epsilon:
        m = (a + b) / 2
        if f(m) == 0:
            return m
        elif f(a) * f(m) < 0:
            b = m
        else:
            a = m
        print(f"Iteration {counter} m = {m}, (a,b) = ({a}, {b}), b-a {b-a}") 
        counter += 1
    return m

m = Bissection3(lambda x : x**2 - 2, 0, 2, 1e-4)
```

## 2. Newton's method:
$$
x_{i+1} = x_i -\frac{f(x_i)}{f'(x_i)}
$$

```python
def Newton1(f, fprime, x0, iMax = 50):
    '''
    Given a function f, its derivative and an initial guess,
    solve the equation f(x) = 0 using Newton method
    '''
    i = 0
    xold = x0
    while i <= iMax:
        xnew = xold - f(xold) / fprime(xold)
        i += 1
        xold = xnew
        print(f"iteration {i}: x_i = {xnew}")
    return xnew

def f(x):
    return x**2-2
def fprime(x):
    return 2*x
```

```python
root = Newton1(f, fprime, 1, 4)
print(f"error: {abs(root-math.sqrt(2))}")
```

```python
root = Newton1(f, fprime, 0, 4)
print(f"error: {abs(root-math.sqrt(2))}")
```

```python
def Newton2(f, fprime, x0, iMax, atol):
    '''
    Given a function f, its derivative and an initial guess,
    solve the equation f(x) = 0 using Newton method
    '''
    i = 0
    xold = x0
    while i <= iMax and abs(f(xold)) >= atol:
        # checking if derivative is zero, to 
        # avoid division by zero error
        if fprime(xold) != 0:
            xnew = xold - f(xold) / fprime(xold)
            i += 1
            xold = xnew
            print(f"iteration {i}: x_i = {xnew}, f(x_i) = {f(xnew)}")
        else:
            print(f"Newton method failed at iteration {i}")
            xnew = xold
            break
    return xnew
Newton2(f, fprime, 1, 10, 1e-5)
```

```python
import math
def g(x):
    return math.sin(x**2)
def gprime(x):
    return 2*x * math.cos(x**2)

x0 = 1.36
root  = Newton2(g, gprime, x0, 10, 1e-5)

%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
X = np.linspace(-5,5,1001)
fig,ax = plt.subplots()
ax.plot(X,[g(x) for x in X])
ax.scatter(x0,g(x0), label = 'x0')
ax.scatter(root,g(root), label = 'root')
ax.grid()
ax.legend()
```
