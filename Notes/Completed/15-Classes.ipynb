# Classes in Python

So far we have seen most of Python's built-in classes (or types as we sometimes refer to them.) Most of the time these types will be sufficient for our purposes. However, in the sciences and mathematics, we generally have very specific or niche goals for our programs. In these cases the Python built-in types might not quite work. Luckily Python gives us the capability to make our own types called **classes**. 

We can start creating a class by using the $\mathtt{class}$ keyword. We need to give any new class a name. It is convention that we use upper case names for classes.

```python
class NewClass:
    pass # we need to say pass for our empty class otherwise Python will be angry with us.
```

In order to make **objects** in our class, we need an *initialization* function. This function has the special name $\mathtt{\_\_init\_\_()}$. The init function takes at least one argument, which we usually call $\mathtt{self}$. Otherwise, the init function can take any information we want our **object** of this class to have.

```python
class NewClass:
    def __init__(self,info1,info2,info3):
        self.info1 = info1
        self.info2 = info2
        self.info3 = info3
```

At this point we can now instantiate an object in the class $\mathtt{NewClass}$, in the following way:

```python
a = NewClass("first property","second property","third interesting property")
```

However, we can't yet look at $a$ quite how we'd want to.

```python
print(a)
```

In order to fix this, we will have to tell Python how it should display elements of our new class. This is done with the $\mathtt{\_\_str\_\_}$ function. The str function takes as input one variable (self), and outputs the string representation of this object so that it can be printed to the screen.

For example:

```python
class NewClass:
    def __init__(self,info1,info2):
        self.info1 = info1
        self.info2 = info2
    def __str__(self):
        return f"Here are the two facts about your object\n Fact 1:{self.info1},\n Fact 2:{self.info2}."
```

Now we can print out an object of the class $\mathtt{NewClass}$.

```python
a = NewClass("a is a really cool object","a has type NewClass")
print(a)
```

# A mathematical example - Fractions!

In mathematics, we often want to deal with fractions instead of their corresponding decimal expansions. However, in Python when we write an expression such as $\mathtt{3/5}$ it is immediately evaluated to its decimal expansion as a floating point number. To get around this we can make our own class implementing fractions.

To start we should name the class and choose how we are going to construct the fraction.

```python
class Frac:
    def __init__(self,denominator,numerator):
        self.denominator = denominator
        self.numerator = numerator
```

Next, we want to print out fractions in some reasonable way. So, we will have to specify the output of the \_\_str\_\_ function. 

```python
class Frac:
    def __init__(self,numerator,denominator):
        self.numerator = numerator
        self.denominator = denominator
    def __str__(self):
        return f"{self.numerator}/{self.denominator}"
```

```python
a = Frac(1,2)
print(a)
```

Next, we should understand when two objects of this class are the same. When dealing with fractions, we usually want to consider $1/2$ and $2/4$ to be the same fraction. In order to do this we will have to implement a reduced form method (for this we will use the gcd function from the math module):

```python
import math

class Frac:
    def __init__(self,numerator,denominator):
        self.numerator = numerator
        self.denominator = denominator
    def __str__(self):
        return f"{self.numerator}/{self.denominator}"
    def reduce(self):
        a = self.numerator
        b = self.denominator
        d = math.gcd(a,b)
        return Frac(a//d,b//d)
```

```python
a = Frac(2,4)
print(a)
b = a.reduce()
print(b)
```

We can now tell our class when two objects are equal.

# Dual numbers

## William Clifford (1873)

**Dual numbers** like the real numbers with one extra number $\epsilon$ the infinitesimal. It satisfies two properties:
- $\epsilon\not=0$
- $\epsilon^2=0$

### Addition

$$(1+7\epsilon)+(2+5\epsilon)=3+12\epsilon$$

### Multiplication

$$(1+7\epsilon)(2+5\epsilon)=2+5\epsilon+14\epsilon +35\epsilon^2=2+19\epsilon$$

$$(a+b\epsilon)(c+d\epsilon)=ac+(ad+bc)\epsilon$$

```python
# implement your class here

class Dual:
    def __init__(self,real,dual):
        self.real = real
        self.dual = dual
    def __str__(self):
        return f"{self.real}+{self.dual}E"
    def __add__(self,other):
        if (isinstance(other,Dual)):
            real = self.real + other.real
            dual = self.dual + other.dual
            return Dual(real,dual)
        return Dual(self.real+other,self.dual)
    __radd__ = __add__
    def __mul__(self,other):
        if (isinstance(other,Dual)):
            a = self.real
            b = self.dual
            c = other.real
            d = other.dual
            real = a*c
            dual = b*c + a*d
            return Dual(real,dual)
        return Dual(self.real*other,self.dual*other)
    __rmul__ = __mul__
```

# Why are dual numbers interesting?

The dual numbers have one nice property that we are interested in. They let us implement differentiation in a very easy way.

**Fact:** Given a function $f(x)$, the derivative of $f(x)$ can be computed using dual numbers as follows:
- Plug $(x+\epsilon)$ into $f(x)$.
- Expand.
- Rearrange so that $f(x+\epsilon)$ has the form 
$$\text{stuff involving x}+\text{stuff involving }\epsilon$$

The first term is the original function and the second term is the derivative!

For example, if $f(x)=x^2+4$, we know that $f'(x)=2x$, we can compute this using dual numbers as follows:
\begin{align*} 
    f(x+\epsilon)&=(x+\epsilon)^2+4,\\
    &=x^2+2x\epsilon+\epsilon^2+4,\\
    &=x^2+4+2x\epsilon+0,\\
    &=(x^2+4)+(2x\epsilon).
\end{align*}

We now see that the term in front of $\epsilon$ is the derivative.

# Iteration

Throughout this semester we have labeled certain types **iterable**. We have seen that lists, sets, tuples, and dictionaries are all iterable. Meaning that we can iterate over them with keywords like "for". But what does this really mean? 

A type in Python is said to be **iterable** if it implements two special functions as methods within its class definition: \_\_iter\_\_ and \_\_next\_\_.

```python
class Counter:
    def __init__(self,low,high):
        self.current = low 
        self.low = low
        self.high = high
    def __iter__(self):
        return self
    def __next__(self):
        if self.current <= self.high:
            current = self.current
            self.current+=1
            return current
        raise StopIteration

a = Counter(2,5)
for i in a:
    print(i)
```
