## Recall the Fibonacci sequence

The Fibonacci sequence is defined as 
$$F(0)=1, F(1)=1, F(n)=F(n-1)+F(n-2)$$

We tried earlier to compute the Fibonacci sequence using while loops. 

We would generally end up with something like the below program:

```python
# Compute the n-th fibonacci number
n = 10
Fnmin1 = 1
Fnmin2 = 1
Fn = 1

# Ctr
i = 1
while i < n:
    if i == 1 or i == 2:
        print(f"F({i}) = 1")
        i+=1
    else:
        Fn = Fnmin1 + Fnmin2
        print(f"F({i}) = {Fn}")
        Fnmin2 = Fnmin1
        Fnmin1 = Fn
        i+=1

```

### Exercise 

Write a function Fib(n) which takes as input a positive integer $n$ and returns the $n$-th Fibonacci number.

```python
def Fib(n : int) -> int:
    # Compute the n-th fibonacci number
    Fnmin1 = 1
    Fnmin2 = 1
    Fn = 1
    i = 1
    while i <= n:
        if i == 1 or i == 2:
            Fn = 1
            i +=1
        else:
            Fn = Fnmin1 + Fnmin2
            Fnmin2 = Fnmin1
            Fnmin1 = Fn
            i+=1
    return Fn
Fib(4)
    
```

```python
def Fib2(n : int) -> int:
    if n == 1 or n == 2:
        return 1
    else:
        return Fib2(n-1) + Fib2(n-2)

print(Fib2(3))
```

# Functions (part IV) - recursion

In mathematics, many formulas are *recursive*. Typically, the $n^{th}$ step of an algorithm or a sequence is defined in terms of the $(n-1)^{th}$.
A classical example is the factorial:

$$
n! = \begin{cases}
1 & \text{ if } n = 0,\\
(n-1)! & \text{ otherwise.}
\end {cases}
$$

Noticing that this gives us $n! = 1. 2. 3. \dots . (n-1) . n$, we can implement the factorial function using a for loop:

```python
def factorial1(n: int) -> int:
    ''' Compute n! using a for loop'''
    # products start with 1
    fact = 1 
    for i in range(1, n+1):
        fact *= i
        # fact = 1*1
        # fact = 1*1*2
        # fact = 1*1*2*3
        # ...
        # fact = 1*1*2*3...*n-1*n
    return fact
```

```python
for i in range(8):
    print(f'{i}! = {factorial1(i)}')
```

We could however have implemented the factorial function in a much more natural way which aligns itself more closely with the recursive mathematical definiton.

```python
def factorial2(n):
    # base case
    if n == 0:
        return 1
    # inductive step (recursive step)
    else:
        return n * factorial2(n-1)
## We could have just done return n * factorial2(n-1)
```

```python
for i in range(8):
    print(f'{i}! = {factorial2(i)}')
```

Note how `factorial2` is called inside itself. This can be visualized this way:

```python
def factorial2(n):
    if n == 0:
        return 1
    else:
        print(f'calling factorial2({n-1})')
        fact = factorial2(n-1)
        # recursion function call
        print(f'returning factorial2({n-1})')
        return n * fact
```

```python
factorial2(5)
```

## Application - Sorting Algorithms

- One of the most common and important family of algorithms in computer science (and mathematics for that matter) are sorting algorithms. 
- The job of a sorting algorithm is as follows: Take in an ordered structure with elements that may be compared under some order e.g. a list of integers, and sort this structure into an ordered state.
- Many of these algorithms are written recursively. We will see two here!

### The Bubble Sort algorithm
- **Idea:** Move along the list comparing each adjacent pair of elements. If they are locally in the wrong order swap them. Repeat until no more swaps are needed.

```python
# [4,3,2,1]
# [3,4,2,1]
# [3,2,4,1]
# [3,2,1,4]

# [2,3,1,4]
# [2,1,3,4]

# [1,2,3,4]

```

```python
def bubbleSort(l : list) -> list:
    n = len(l)
    swapped = False
    for i in range(1,n):
        if l[i-1] > l[i]: # checking if entries should be swapped
            l[i-1],l[i] = l[i],l[i-1] # swap
            swapped = True
    if swapped == False:
        return l
    else:
        return bubbleSort(l)
```

## Application: the quicksort algorithm

The quicksort algorithm (Tony Hoare, 1959) is a "divide-and-conquer" algorithm for sorting a list.
Given a list L, we pick a "pivot" value p in L and split L into three parts: 
* Ll: the numbers in L that are less than p
* Le: the numbers in L that are equal to p
* Lu: the numbers in L that are greater than p

We then apply quicksort to Ll and Lu and return 
Ll + Le + Lu

It is very easy to implement using recursion

```python
def quicksort(L):
    if len(L) <= 1:
        return L
    p = L[len(L)//2]
    Ll = []
    Le = []
    Lu = []
    for l in L:
        if l < p:
            Ll.append(l)
        elif l == p:
            Le.append(l)
        else:
            Lu.append(l)
    print(f'pivot: {p}, {L} = {Ll} + {Le} + {Lu}')
    return quicksort(Ll) + Le + quicksort(Lu) 
```

```python
quicksort([1,3,6,-1,2,5,1])
```
